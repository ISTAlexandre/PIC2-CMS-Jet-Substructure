```python
import ROOT
import numpy as np
import pandas as pd

CSV = "../csv/merged_ML_ch2-pb80.csv"
pandas_df = pd.read_csv(CSV)
print(f"Number of events in CSV: {len(pandas_df)}")
```

```python
hist = ROOT.TH1F("hist_jet_pt", "Jet pT distribution;Jet pT [GeV];Events", 100, 100, 400)
for row in pandas_df.itertuples(index=False):
    jet_pt = row.Jet_pt
    hist.Fill(jet_pt)
```

```python
canvas = ROOT.TCanvas("canvas", "dpsi12 distribution for channel2 = 1", 800, 600)
hist.GetXaxis().SetRangeUser(700, 2000)
hist.SetLineColor(ROOT.kRed)
hist.Draw()
canvas.Draw()
```
