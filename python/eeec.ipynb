```python
import ROOT
import numpy as np


file_path = "../root_ML/merged_ML-600to800.root"

hist_names = [
    "lund_eeec_dpsi",
    "lund_eeec_dpsi_ch0",
    "lund_eeec_dpsi_ch1",
    "lund_eeec_dpsi_ch2",
    "lund_eeec_dpsi_ch3",
    "lund_eeec_dpsi_ch4",
]


def mean_cos2dpsi(hist):
    numerator = 0.0
    denominator = 0.0

    nbins = hist.GetNbinsX()

    for ibin in range(1, nbins + 1):
        dpsi = hist.GetBinCenter(ibin)
        weight = 8*hist.GetBinContent(ibin)

        numerator += weight * np.cos(2.0 * dpsi)
        denominator += weight

    if denominator == 0.0:
        return np.nan, numerator, denominator

    return numerator / denominator, numerator, denominator


f = ROOT.TFile.Open(file_path, "READ")
if not f or f.IsZombie():
    raise RuntimeError(f"Could not open {file_path}")

print()
print("Histogram <cos(2 dpsi)> results")
print("--------------------------------")

results = {}
canvases = []

for name in hist_names:
    hist = f.Get(name)
    canvas = ROOT.TCanvas(name, name, 800, 600)
    hist.Draw()
    canvases.append(canvas)
    canvas.SaveAs(f"../{name}.png")

    if not hist:
        print(f"{name:25s} : not found")
        continue

    value, numerator, denominator = mean_cos2dpsi(hist)
    results[name] = value

    print(f"{name:25s} : <cos(2dpsi)> = {value:.8f}   integral = {denominator:.8e}")

f.Close()
```
