```python
import ROOT
import numpy as np

file = ROOT.TFile.Open("../root_ML/merged_ML-600to800.root")
tree = file.Get("jetTree")

n_events = tree.GetEntries()
print(f"Number of events: {n_events}")
```

```python
coords = []
hist_channel2_dpsi12_1 = ROOT.TH1F("hist_channel2_dpsi12_1", "dpsi12 distribution channel2 = 1", 100, 0, np.pi)
hist_channel2_dpsi12_1.GetXaxis().SetTitle("dpsi12")
hist_channel2_dpsi12_1.GetYaxis().SetTitle("Counts")

hist_jet_pt = ROOT.TH1F("hist_jet_pt", "Jet pT distribution", 100, 700, 2000)
hist_jet_pt_ch1 = ROOT.TH1F("hist_jet_pt_ch1", "Jet pT distribution for channel2 = 1", 100, 700, 2000)
hist_jet_pt_ch2 = ROOT.TH1F("hist_jet_pt_ch2", "Jet pT distribution for channel2 = 2", 100, 700, 2000)
hist_jet_pt_ch3 = ROOT.TH1F("hist_jet_pt_ch3", "Jet pT distribution for channel2 = 3", 100, 700, 2000)
hist_jet_pt_ch4 = ROOT.TH1F("hist_jet_pt_ch4", "Jet pT distribution for channel2 = 4", 100, 700, 2000)
hist_jet_pt_ch0 = ROOT.TH1F("hist_jet_pt_ch0", "Jet pT distribution for channel2 = 0", 100, 700, 2000)

hist_p3_n_all = ROOT.TH1F("hist_p3_n_all", "p3_n_all", 10, 0, 40)
hist_p3_n_all_ch1 = ROOT.TH1F("hist_p3_n_all_ch1", "p3_n_all ch1", 10, 0, 40)
hist_p3_n_all_ch2 = ROOT.TH1F("hist_p3_n_all_ch2", "p3_n_all ch2", 10, 0, 40)
hist_p3_n_all_ch3 = ROOT.TH1F("hist_p3_n_all_ch3", "p3_n_all ch3", 10, 0, 40)
hist_p3_n_all_ch4 = ROOT.TH1F("hist_p3_n_all_ch4", "p3_n_all ch4", 10, 0, 40)
hist_p3_n_all_ch0 = ROOT.TH1F("hist_p3_n_all_ch0", "p3_n_all ch0", 10, 0, 40)

hist_pt_list = [hist_jet_pt_ch0, hist_jet_pt_ch1, hist_jet_pt_ch2, hist_jet_pt_ch3, hist_jet_pt_ch4]
hist_p3_n_all_list = [hist_p3_n_all_ch0,hist_p3_n_all_ch1,hist_p3_n_all_ch2,hist_p3_n_all_ch3,hist_p3_n_all_ch4]

for n_entry in range(n_events):
    tree.GetEntry(n_entry)
    n_jets = len(tree.jet_pt)

    for jet_i in range(n_jets):

        channel2 = tree.lund_secondary_idx_sd[jet_i]
        channel1 = tree.lund_primary_idx_sd[jet_i]
        max_kt_index1 = int(tree.lund_max_kt_sd[jet_i])
        max_kt_index2 = int(tree.lund_max_kt_secondary_sd[jet_i])
        dpsi12 = tree.lund_psi12_sd[jet_i]

        p3_n_all = tree.lund_p3_n_all[jet_i]

        if channel2 == -1 or max_kt_index1 < 0 or max_kt_index2 < 0 or channel1 == -1:
            continue

        coords.append((channel1, channel2))

        if channel2 == 1:
            hist_channel2_dpsi12_1.Fill(dpsi12)

        jet_pt = tree.jet_pt[jet_i]
        
        hist_jet_pt.Fill(jet_pt)
        hist_p3_n_all.Fill(p3_n_all)

        hist_pt_list[channel2].Fill(jet_pt)
        hist_p3_n_all_list[channel2].Fill(p3_n_all)
```

```python
total_coords = len(coords)
for i in range(5):
    for j in range(5):
        count = sum(1 for c in coords if c == (i, j))
        print(f"Channel 1: {i}, Channel 2: {j}, Count: {count}, Percentage: {count/total_coords:.2%};")
```

```python
canvas = ROOT.TCanvas("canvas", "dpsi12 distribution channel2 = 1", 800, 600)
hist_channel2_dpsi12_1.SetLineColor(ROOT.kBlue)
hist_channel2_dpsi12_1.Draw()
canvas.Draw()
canvas.SaveAs("../imgs/frame.png")
```

```python
canvas_jet_pt = ROOT.TCanvas("canvas_jet_pt", "Jet pT distribution", 800, 600)

hist_jet_pt.SetLineColor(ROOT.kRed)
hist_jet_pt.Scale(1.0 / hist_jet_pt.Integral())

hist_jet_pt_ch0.SetLineColor(ROOT.kGreen)
hist_jet_pt_ch1.SetLineColor(ROOT.kBlue)
hist_jet_pt_ch2.SetLineColor(ROOT.kPink)
hist_jet_pt_ch3.SetLineColor(ROOT.kOrange)
hist_jet_pt_ch4.SetLineColor(ROOT.kYellow)

hist_jet_pt_ch0.Scale(1.0 / hist_jet_pt_ch0.Integral())
hist_jet_pt_ch1.Scale(1.0 / hist_jet_pt_ch1.Integral())
hist_jet_pt_ch2.Scale(1.0 / hist_jet_pt_ch2.Integral())
hist_jet_pt_ch3.Scale(1.0 / hist_jet_pt_ch3.Integral())
hist_jet_pt_ch4.Scale(1.0 / hist_jet_pt_ch4.Integral())

hist_jet_pt.Draw("HIST")
hist_jet_pt_ch0.Draw("HIST SAME")
hist_jet_pt_ch1.Draw("HIST SAME")
hist_jet_pt_ch2.Draw("HIST SAME")
hist_jet_pt_ch3.Draw("HIST SAME")
hist_jet_pt_ch4.Draw("HIST SAME")

legend = ROOT.TLegend(0.7, 0.7, 0.9, 0.9)
legend.AddEntry(hist_jet_pt, "All channels", "l")
legend.AddEntry(hist_jet_pt_ch0, "Channel 2 = 0", "l")
legend.AddEntry(hist_jet_pt_ch1, "Channel 2 = 1", "l")
legend.AddEntry(hist_jet_pt_ch2, "Channel 2 = 2", "l")
legend.AddEntry(hist_jet_pt_ch3, "Channel 2 = 3", "l")
legend.AddEntry(hist_jet_pt_ch4, "Channel 2 = 4", "l")
legend.Draw()

canvas_jet_pt.Draw()
```
